```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ajout de src au PYTHONPATH
sys.path.append(str(Path("..").resolve()))
from src.data_inspection import load_dataset
from src.data_cleaning import clean_dataset, detect_outliers_iqr, generate_comparison_table

DATA_PATH = Path("../data/EV_Predictive_Maintenance_Dataset_15min.csv")
df_raw = load_dataset(DATA_PATH)
print("Dimensions brutes :", df_raw.shape)
```

## 1. Exécution du nettoyage et validation

```python
df_clean, meta = clean_dataset(df_raw)
print("Rapport de nettoyage :", meta)

# Tableau comparatif avant/après
comp_table = generate_comparison_table(df_raw, df_clean)
comp_table
```

## 2. Détection des valeurs extrêmes (IQR)
Analyse de Tukey sur les variables opérationnelles du véhicule électrique.

```python
outliers_df = detect_outliers_iqr(df_clean)
outliers_df
```

## 3. Visualisation des distributions et valeurs extrêmes
Examen des variables clés de surveillance thermique et mécanique.

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

sns.boxplot(y=df_clean["Motor_Temperature"], ax=axes[0], color="#e74c3c")
axes[0].set_title("Température Moteur (°C)", fontsize=12, fontweight="bold")
axes[0].set_ylabel("Degrés Celsius")

sns.boxplot(y=df_clean["Motor_Vibration"], ax=axes[1], color="#f39c12")
axes[1].set_title("Vibration Moteur (mm/s)", fontsize=12, fontweight="bold")
axes[1].set_ylabel("Vibration (mm/s)")

sns.boxplot(y=df_clean["Battery_Temperature"], ax=axes[2], color="#3498db")
axes[2].set_title("Température Batterie (°C)", fontsize=12, fontweight="bold")
axes[2].set_ylabel("Degrés Celsius")

plt.tight_layout()
plt.show()
print("Interprétation : Les valeurs extrêmes observées (moteur jusqu'à 100°C, vibration jusqu'à 2 mm/s) sont physiquement cohérentes et représentent les signatures critiques annonciatrices de panne. Elles ne doivent pas être éliminées.")
```
